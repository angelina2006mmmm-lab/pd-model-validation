# EDA — Synthetic Credit Risk Dataset
Quick exploratory analysis for the PD validation pet project.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(root / 'data' / 'credit_risk_synthetic.csv')
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
print('Default rate:', df['default_12m'].mean())
print('\nMissing values:')
display(df.isna().sum().sort_values(ascending=False))
```

```python
df.groupby('sector_risk')['default_12m'].agg(['count', 'mean']).plot(kind='bar')
plt.ylabel('Observed default rate')
plt.tight_layout()
plt.show()
```
